```python
import glob, os
cands = sorted(glob.glob("/kaggle/input/**/adapter_config.json", recursive=True))
assert cands, "adapter_config.json not found under /kaggle/input"
model_dir = os.path.dirname(cands[0]); print("adapter dir:", model_dir)
for fn in ["adapter_config.json","adapter_model.safetensors"]:
    assert os.path.exists(os.path.join(model_dir, fn)), f"missing {fn}"
rc = os.system(f'cd "{model_dir}" && zip -j /kaggle/working/submission.zip adapter_config.json adapter_model.safetensors')
print("zip rc:", rc, "size MB:", os.path.getsize("/kaggle/working/submission.zip")/1e6)
```
